# Projeto queimadas
1. Listar arquivos dentro da pasta raw_data
2. Consolidar em apenas 1 dataset as bases
3. Salvar na Camada bronze no formato Delta

```python
# Importando bibliotecas
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
# Listando os arquivos dentro da pasta ´raw_data´
lista_arquivos = dbutils.fs.ls('/Volumes/proj_queimadas/default/raw_data')
lista_arquivos
```

```python
# Acessando o caminho(path) dentro da variável lista_arquivos
lista_arquivos_path = [arquivos.path for arquivos in lista_arquivos]
lista_arquivos_path
```

```python
# Consolidando todos os arquivos da pasta em apenas 1 DataFrame

# for arquivo in lista_arquivos_path:
#     df_temp = spark.read.csv(arquivo,header=True)
#     df = df_temp.union(df_temp)
#     print(f'{arquivo} concluído com sucesso')

df = None

for arquivo in lista_arquivos_path:
    df_temp = spark.read.csv(arquivo, header=True)
    df = df_temp if df is None else df.unionByName(df_temp, allowMissingColumns=True)
    print(f'{arquivo} concluído com sucesso')
```

```python
# Salvando o DataFrame no formato Delta na Camada Bronze
(
    df.write
    .format('Delta')
    .mode('overwrite')
    .option("overwriteSchema", "true")
    .saveAsTable('proj_queimadas.bronze.bronze_queimadas')
    
)
```
