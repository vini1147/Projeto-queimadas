```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
```

```python
%sql
-- Analisando alguns dados com SQL
select * from proj_queimadas.bronze.bronze_queimadas
```

#### Lendo a tabela bronze_queimadas na camada `bronze`

```python

df_queimadas = (
    spark.read.table('proj_queimadas.bronze.bronze_queimadas')
)
```

Substituindo valores nas colunas `numero_dias_sem_chuva`,`precipitacao`,`risco_fogo` e `frp` de "-999.0" para "0.0"

```python
df_queimadas = df_queimadas.withColumns(
    {
        'numero_dias_sem_chuva':when(col('numero_dias_sem_chuva').startswith('-'),'0.0').otherwise(col('numero_dias_sem_chuva')),
        'precipitacao':when(col('precipitacao').startswith('-'),'0.0').otherwise(col('precipitacao')),
        'risco_fogo':when(col('risco_fogo').startswith('-'),'0.0').otherwise(col('risco_fogo')),
        'frp':when(col('frp').startswith('-'),'0.0').otherwise(col('frp'))
    }
)
```

#### Colocando colunas no padrão primeira letra em maiúscula

```python
df_queimadas.withColumns(
    {
        'estado':initcap(col('estado')),
        'municipio':initcap(col('municipio'))
    }
).show(5)
```
